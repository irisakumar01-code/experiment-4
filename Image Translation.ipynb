# **Image Translation**

Write a Python program using OpenCV to perform image translation on the **"Given image"**.

The program should:

1. Import `cv2`, `numpy`, and `matplotlib.pyplot`.
2. Read the image **"given** in color mode.
3. Create a **translation matrix** that moves the image  
   **50 pixels to the right** and **80 pixels down**.
4. Apply the translation
5. Display the following using a **1 × 2 layout** in Matplotlib:  
   - **Original Image**  
   - **Translated Image**


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt 

```

```python
img = cv2.imread("baseball.jpg")

```

```python
M = np.float32([[1, 0, 50],
                [0, 1, 80]])
```

```python
translated = cv2.warpAffine(img, M, (500, 500))

```

```python
plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original Image")

plt.subplot(1, 2, 2)
plt.imshow(translated)
plt.title("Translated Image")

plt.show()
```

# **Image Shearing**

Write a Python program using OpenCV to perform image shearing on the **given image**.

The program should:

1. Create a **shearing transformation matrix** to apply:
   - **Horizontal shearing**  
   - **Vertical shearing**
2. Apply the shearing transformations.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **Horizontally Sheared Image**  
   - **Vertically Sheared Image**


```python
M1 = np.float32([[1, 0.5, 0], [0, 1, 0]])  # Horizontal
M2 = np.float32([[1, 0, 0], [0.5, 1, 0]])  # Vertical
```

```python
horizontal = cv2.warpAffine(img, M1, (img.shape[1], img.shape[0]))
vertical = cv2.warpAffine(img, M2, (img.shape[1], img.shape[0]))
```

```python
plt.subplot(1, 3, 1)
plt.imshow(img)
plt.title("Original")

plt.subplot(1, 3, 2)
plt.imshow(horizontal)
plt.title("Horizontal Shear")

plt.subplot(1, 3, 3)
plt.imshow(vertical)
plt.title("Vertical Shear")

plt.show()
```

# **Image Rotation**

Write a Python program using OpenCV to perform **image rotation** on the **given image**.

The program should:

1. Rotate the image by:  
   - **45 degrees**  
   - **90 degrees**
2. Use `cv2.getRotationMatrix2D` and `cv2.warpAffine`.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **45° Rotated Image**  
   - **90° Rotated Image**


```python
h, w = img.shape[:2]

M1 = cv2.getRotationMatrix2D((w//2, h//2), 45, 1)
M2 = cv2.getRotationMatrix2D((w//2, h//2), 90, 1)

rot45 = cv2.warpAffine(img, M1, (w, h))
rot90 = cv2.warpAffine(img, M2, (w, h))
```

```python
plt.subplot(1, 3, 1)
plt.imshow(img)
plt.title("Original")

plt.subplot(1, 3, 2)
plt.imshow(rot45)
plt.title("45 Degree")

plt.subplot(1, 3, 3)
plt.imshow(rot90)
plt.title("90 Degree")

plt.show()
```
